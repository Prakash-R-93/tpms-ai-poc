```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Python environment is ready")
```

```python
from pathlib import Path

data_dir = Path("../data/raw/dataset1")

print(data_dir)
```

```python
files = list(data_dir.glob("*.csv"))

for file in files:
    print(file)
```

```python
daily_df = pd.read_csv(data_dir / "daily_user.csv")
heavy_df = pd.read_csv(data_dir / "heavy_user.csv")
moderate_df = pd.read_csv(data_dir / "moderate_user.csv")
rare_df = pd.read_csv(data_dir / "rare_user.csv")
```

```python
print("Daily:", daily_df.shape)
print("Heavy:", heavy_df.shape)
print("Moderate:", moderate_df.shape)
print("Rare:", rare_df.shape)
```

```python
print("Daily columns:")
print(daily_df.columns.tolist())

print("\nHeavy columns:")
print(heavy_df.columns.tolist())

print("\nModerate columns:")
print(moderate_df.columns.tolist())

print("\nRare columns:")
print(rare_df.columns.tolist())
```

```python
schemas_match = (
    list(daily_df.columns) == list(heavy_df.columns) ==
    list(moderate_df.columns) == list(rare_df.columns)
)

print("All schemas identical:", schemas_match)
```
